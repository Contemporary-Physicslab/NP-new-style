# Wat is de onzekerheid?
Hier laten we zien dat de onzekerheid in het gemiddelde van een dataset gelijk is aan de standaard deviatie van het gemiddelde van herhaalde datasets.

```python
import numpy as np
import matplotlib.pyplot as plt

dataset = np.random.normal(0,10,1000)

plt.figure()
plt.plot(dataset)
plt.show()

plt.figure()
plt.hist(dataset)
plt.show()
```

```python
dataset = np.random.normal(0,10,1000)

print('gemiddeld: ', np.mean(dataset), ' onzekerheid: ', np.std(dataset,ddof=1)/np.sqrt(len(dataset)))
```

```python
means = []
for i in range(1000):
    means.append(np.mean(np.random.normal(0,10,1000)))
```

```python
print(np.std(means))
```
